```python
print('Calcule seu IMC')
```

```python
Altura = float(input('Digite sua altura em centímentors: ')) # aguardar usuário digitar algo
```

```python
Peso = float(input('Digite seu peso: ').replace(',', '.')) # aguardar usuário digitar algo
```

```python
# Fórmula IMC
IMC = Peso / ((Altura / 100) * (Altura / 100))

print(f"Seu IMC é: {IMC:.2f}")

if IMC <= 18.5:
  print ('Você está abaixo do peso')
elif IMC > 18.5 and IMC <= 24.9:
  print('Seu peso está normal')
elif IMC > 24.9 and IMC <= 29.9:
  print('Você está com sobrepeso')
elif IMC > 29.9 and IMC <= 34.9:
  print('Você está com Obesidade Grau I')
elif IMC > 34.9 and IMC <= 39.9:
  print('Você está com Obesidade Grau II (Severa)')
else:
  print('Você está com Obesidade Grau III (Mórbida)')
```
